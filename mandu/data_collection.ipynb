```python
import requests

genres = set()

for page in range(3):  # 테스트용
    url = f"https://steamspy.com/api.php?request=all&page={page}"
    data = requests.get(url).json()

    for game in data.values():
        genre_str = game.get('genre')

        if genre_str:
            for g in genre_str.split(','):
                genres.add(g.strip())

print(sorted(genres))
```
