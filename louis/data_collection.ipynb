```python
import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta

# 삼양식품: 003230.KS
ticker = "003230.KS"

end_date = datetime.today()
start_date = end_date - timedelta(days=365 * 3)

df = yf.download(
    ticker,
    start=start_date.strftime("%Y-%m-%d"),
    end=end_date.strftime("%Y-%m-%d"),
    interval="1d",
    auto_adjust=False
)

df.head()
```

```python
df = df.reset_index()

df = df.rename(columns={
    "Date": "date",
    "Open": "open",
    "High": "high",
    "Low": "low",
    "Close": "close",
    "Adj Close": "adj_close",
    "Volume": "volume"
})

df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))
plt.plot(df["date"], df["close"])
plt.title("Samyang Foods Stock Price - 3 Years")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.grid(True)
plt.show()
```

```python
import os
import requests
import xmltodict
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

KAMIS_API_KEY = os.getenv("KAMIS_API_KEY")
KAMIS_API_ID = os.getenv("KAMIS_API_ID")

assert KAMIS_API_KEY, "KAMIS_API_KEY가 .env에 없습니다."
assert KAMIS_API_ID, "KAMIS_API_ID가 .env에 없습니다."

BASE_URL = "http://www.kamis.or.kr/service/price/xml.do"

params = {
    "action": "periodProductList",
    "p_cert_key": KAMIS_API_KEY,
    "p_cert_id": KAMIS_API_ID,
    "p_returntype": "xml",

    # 조회 기간
    "p_startday": "2024-06-01",
    "p_endday": "2024-06-10",

    # 01: 소매, 02: 도매
    "p_productclscode": "01",

    # 쌀 테스트 예시
    "p_itemcategorycode": "100",  # 식량작물
    "p_itemcode": "111",         # 쌀
    "p_kindcode": "01",         # 일반계
    "p_productrankcode": "",     # 등급 코드, 필요 없으면 공백
    "p_countrycode": "",         # 지역 코드, 전국 평균이면 공백
}

res = requests.get(BASE_URL, params=params, timeout=30)

print("status_code:", res.status_code)
print("content_type:", res.headers.get("Content-Type"))
print("response preview:")
print(res.text[:1000])

res.raise_for_status()
```
